# Goal
First, compare word frequencies between two works of Fitzgerald. Compare word frequencies between works of Shelley and Fitzgerald. Then check if there's some words preferred by Fitzgerald that are used less frequently than Shelley. 

# How to Achieve Goal

1. Download and clean the text of each book (remove HTML/markup, lowercase, strip punctuation, remove stop words).
2. Compute word frequency counts for each book.
3. Compare word frequencies between *This Side of Paradise* and *The Great Gatsby* (same author).
4. Compare word frequencies between the Fitzgerald books and *Frankenstein* (different author).
5. Normalize frequencies (count / total words) and find words with a much higher relative frequency in Fitzgerald's works than in Shelley's.
